# 03 · Visualization and Dashboard

This notebook keeps the visual review and dashboard preview together. The complete interactive Panel application remains in `app.py`; this notebook imports and displays that same application rather than copying its implementation.

## 1. Check the saved analysis assets

Run notebooks 01 and 02 first. The dashboard reads the cleaned CSV and the single selected model bundle.

```python
from pathlib import Path
import joblib
import pandas as pd
from IPython.display import display
import panel as pn

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'Datasets').exists():
    PROJECT_ROOT = PROJECT_ROOT / 'Customer_Churn_Prediction'
model_bundle = joblib.load(PROJECT_ROOT / 'selected_model.joblib')
df = pd.read_csv(PROJECT_ROOT / 'Datasets' / 'Telco-Customer-Churn_Cleaned.csv')
print('Customers:', len(df))
print('Selected model:', model_bundle['model_name'])
display(model_bundle['test_metrics'].round(3))
display(model_bundle['feature_importance'].head(10))
```
